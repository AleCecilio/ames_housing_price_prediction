## Engenharia de Features

Este arquivo é utilizado para realizar a **engenharia de features**, etapa em que as variáveis do dataset são transformadas para melhor utilização pelos modelos de machine learning. Nessa fase, são aplicadas transformações como a **criação de variáveis dummies para colunas categóricas**, convertendo dados textuais em representações numéricas.

Optou-se por manter essa etapa em um notebook separado para facilitar a **organização e a revisão das transformações realizadas nos dados** antes da etapa de modelagem.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/processed/Ames_NO_missing_data.csv')
```

### Descrição das Features

Utilizaremos o arquivo **Ames_Housing_Feature_Description.txt** para entender melhor o significado de cada *feature* presente no dataset e o que seus respectivos valores representam. Aqui o foco entender a *feature* *Ms* *SubClass*. 


```python
with open('../data/raw/Ames_Housing_Feature_Description.txt', 'r') as f:
    print(f.read())
```

```python
df['MS SubClass']
```

```python
df['MS SubClass'] = df['MS SubClass'].astype(str)
```

```python
my_object_df = df.select_dtypes(include='object')
my_numeric_df = df.select_dtypes(exclude='object')
```

```python
df_objects_dummies = pd.get_dummies(my_object_df, drop_first=True)
```

```python
df_objects_dummies
```

```python
final_df = pd.concat([my_numeric_df, df_objects_dummies], axis=1)
```
